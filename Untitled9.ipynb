```python
import pandas as pd

df = pd.read_csv("ecommerce_sales_analytics_5000.csv",parse_dates=["order_date"])
```

```python
#verify
df.info()
```

```python
df["payment_type"] = df["payment_method"].map({
    "Card": "Digital",
    "wallet": "Digital",
    "Cod": "Cash"
})
# Explanation: Each value is mapped indvivdually->fast & simple
```

```python
#verify
df["payment_type"].unique()     #check the unique values
```

```python
df[["payment_method","payment_type"]].head()       #compare before vs after
```

```python
df.groupby("payment_type").size()                     #Business sanity check
```

```python
#apply() - create revenue category
# apply() - create revenue category
def revenue_category(row):
    if row["revenue"] > 500:
        return "High"
    elif row["revenue"] >= 200:
        return "Medium"
    else:
        return "Low"

df["revenue_category"] = df.apply(revenue_category, axis=1)
#why apply?  decision depends on multiple columns.
        
        
```

```python
#verify
df["revenue_category"].value_counts()
```

```python
df[["revenue","revenue_category"]].sample(10)
#Revenue  > 500 high
#Between  200-500 Medium
# <200 Low
```

```python
#pipe()  - Method chaining
def add_profit(data):
    data["profit"] = data["revenue"]*0.20
    return data

def add_net_revenue(data):
    data["net_revenue"] = data["revenue"] - data["profit"]
    return data

df = (
    df
    .pipe(add_profit)
    .pipe(add_net_revenue)
)
```

```python
#verfiy
df[["revenue", "profit", "net_revenue"]].head()   #check new column
```

```python
                                               #Topic 67 | Outlier Detection Using Statistical Rules
```

```python
import pandas as pd

df = pd.read_csv("ecommerce_sales_analytics_5000.csv",parse_dates=["order_date"])
```

```python
# 1. Calculate Q1 (25th percentile) and Q3 (75th percentile)
Q1 = df["revenue"].quantile(0.25)
Q3 = df["revenue"].quantile(0.75)

# 2. Calculate IQR
IQR = Q3 - Q1

# 3. Define Lower and Upper Bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# 4. Identify Outliers
outliers_iqr = df[
    (df["revenue"] < lower_bound) | (df["revenue"] > upper_bound)
]

#IQR (interquartile range) : measures the middle 50% of dats. robust for skewed distribution (like revenue)
```

```python
outliers_iqr.shape    #Verified
```

```python
outliers_iqr[["revenue", "product_category"]].head()  #inspect
```

```python
df = df[(df["revenue"] >= lower_bound) & (df["revenue"] <= upper_bound)]
df.shape
```

```python
# Z-score Method
import numpy as np
import pandas as pd
from scipy.stats import zscore

df = pd.read_csv("ecommerce_sales_analytics_5000.csv",parse_dates=["order_date"])
df["revenue_z"] = zscore(df["revenue"])
outliers_z = df[df["revenue_z"].abs() > 3]

#Z-score show: how many standard deviation a value is from the mean
# Rule:|z|>3  outlier
```

```python
outliers_z.shape   #verification
```

```python
df = df[df["revenue_z"].abs() <=3]
df.shape
```
