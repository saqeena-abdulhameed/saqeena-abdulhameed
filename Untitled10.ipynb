```python
                                                    #Topic 71 | Line & Scatter Plots Basics
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("ecommerce_sales_analytics_5000.csv",parse_dates=["order_date"])
df = df.sample(n=100)
```

```python
# Line Plot  - Mothly Revenue Trend
monthly_revenue = (
    df
    .set_index("order_date")["revenue"]
    .resample("ME")
    .sum()
)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(monthly_revenue.index, monthly_revenue.values,linewidth=2)

ax.set_title("Monthly Revenue Trend")
ax.set_xlabel("Month")
ax.set_ylabel("Total Revenue")

#Explonation: Line connect revenue over time and shows growth or decline clearly
```

```python
#Scatter Plot - Discount VS  Revenue
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["unit_price"], df["revenue"], alpha=0.5)

ax.set_title("Unit Price VS Revenue")
ax.set_xlabel("unit Price")
ax.set_ylabel("Revenue")

#Each point = one order and Transparency (alpha) hepls with overlapping points
```

```python
                                    #Topic 72 | Bar Charts in Matplotlib - Simple, Grouped & Stacked
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("ecommerce_sales_analytics_5000.csv",parse_dates=["order_date"])
```

```python
# Simple Bar Chart - Revenue by Product Category
category_revenue = df.groupby("product_category")["revenue"].sum()

fig , ax = plt.subplots()
ax.bar(category_revenue.index, category_revenue.values)

ax.set_title("Total Revenue by product Category")
ax.set_xlabel("Category")
ax.set_ylabel("Total Revenue")
```

```python
# Group Bar Chart - Revennue by Category & Region
grouped = df.groupby(["product_category", "region"])["revenue"].sum().unstack()

grouped.plot(kind="bar", figsize=(10, 4))
plt.title("Revenue by Category and Region")
plt.ylabel("Revenue")
plt.show()
```

```python
#stock Bar Chart 
grouped.plot(kind="bar", stacked=True, figsize=(8, 4))
plt.title("Stacked Revenue by Category and Region")
plt.ylabel("Revenue")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

group_a = np.random.normal(70, 10, 1000)
group_b = np.random.normal(80, 12, 1000)

plt.hist(group_a, bins=30, density=True,
         alpha=0.6, label="Group A")

plt.hist(group_b, bins=30, density=True,
         alpha=0.6, label="Group B")

plt.xlabel("score")
plt.ylabel("Density")
plt.title("Distribution")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("ecommerce_sales_analytics_5000.csv",parse_dates=["order_date"])
df = df.sample(n=1000)
```

```python
#Histogram - Revenue Distribution
fig, ax = plt.subplots(figsize=(12, 4))
ax.hist(df["revenue"], bins=30)

ax.set_title("Revenue Distritbution")
ax.set_xlabel("Revenue")
ax.set_ylabel("Frequencey")
```

```python
#Density Plot
fig, ax = plt.subplots(figsize=(12, 4))
ax.hist(df["revenue"],bins=30, density=True, alpha=0.6)

ax.set_title("Revenue Density Distribution")
ax.set_xlabel("Reveune")
ax.set_ylabel("Density")
```

```python
#overlay two histograms (comparison) - Revenue vs discounted
discounted_revenue = df["revenue"] * (1 - df["discount"])

fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(df["revenue"], bins=30, alpha=0.6, label="Original Revenue")
ax.hist(discounted_revenue, bins=30, alpha=0.6, label="Discounted Revenue")

ax.legend()
ax.set_title("Revenue Comparison")
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

fig, axs = plt.subplots(2, 2, figsize=(8, 6))

#Line plot
axs[0, 0].plot(x, y, color="b")
axs[0, 0].set_title("Line Plot")

# Scatter plot
axs[0, 1].scatter(x, y, color="r")
axs[0, 1].set_title("Scatter Plot")

#Histogram
axs[1, 0].hist(np.random.randn(200), bins=20, color="g")
axs[1, 0].set_title("Histrogram")

# Bar Chart
categories = ["A","B","C"]
values = [5, 3, 7]
axs[1, 1].bar(categories, values, color="m")
axs[1, 1].set_title("Bar Chart")

plt.tight_layout()
plt.show()
```

```python
                                        #Topic 75 | Customizing & Exporting Matplotlib Visualizations
```

```python
x = np.linspace(0, 10, 50)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(6,4))
ax.plot(x, y, label="sin(x)", color="b", linewidth=2)

#Customer ticks & rotation
ax.set_xticks(range(0, 11, 2))
plt.xticks(rotation=30)

#legned placement
ax.legend(loc="upper right")

#Label & title
ax.set_title("customizes sine wave pplot")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")

# Save polished figure
plt.tight_layout()
plt.savefig("sine_plot.png", dpi=300, format="png")
plt.show()
```
